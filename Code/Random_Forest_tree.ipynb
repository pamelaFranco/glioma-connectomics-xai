```python
# ==========================================
# CELL 2: Upload file, render with LaTeX font, and download
# ==========================================
from google.colab import files
import os
import re

print("Please upload your .dot (or .txt) file containing the tree structure:")
uploaded = files.upload()

# Get uploaded filename
input_filename = list(uploaded.keys())[0]

# Read uploaded content
with open(input_filename, 'r', encoding='utf-8') as f:
    dot_content = f.read()

# 1. Replace Arial with Latin Modern / LaTeX style typography
dot_content = dot_content.replace('fontname="Arial"', 'fontname="Latin Modern Math"')
dot_content = re.sub(r'fontname\s*=\s*"[^"]+"', 'fontname="Latin Modern Math"', dot_content)

# 2. Clean feature labels: Replace underscores with spaces for cleaner LaTeX-style rendering
dot_content = re.sub(r'([a-zA-Z0-9])_([a-zA-Z0-9])', r'\1 \2', dot_content)

# 3. Save modified DOT file with high DPI setting
modified_dot_filename = "tree_latex.dot"
with open(modified_dot_filename, 'w', encoding='utf-8') as f:
    # Inject dpi=300 at the top of the graph for publication quality
    dot_content_with_dpi = dot_content.replace('digraph Tree {', 'digraph Tree {\n  dpi=300;')
    f.write(dot_content_with_dpi)

# Render to high-resolution PNG and vector PDF
output_png = "Supplementary_Figure_3.png"
output_pdf = "Supplementary_Figure_3.pdf"

os.system(f"dot -Tpng {modified_dot_filename} -o {output_png}")
os.system(f"dot -Tpdf {modified_dot_filename} -o {output_pdf}")

print("\nImage rendered successfully with LaTeX-style fonts!")

# Automatically trigger download for PNG
files.download(output_png)

# Uncomment the line below if you also want to download the vector PDF version for journal submission:
# files.download(output_pdf)
```
